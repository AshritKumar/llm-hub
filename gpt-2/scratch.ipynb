```python
from gpt2_model import GPT2
from dataclasses import dataclass
from torchinfo import summary
```

```python
@dataclass
class GPTConfig:
    context_len  = 1024 # aka block_size, sequence_len, etc..
    batch_size = 32
    emb_dim = 768
    n_heads = 12
    n_transformer_blocks = 12
    dropout = 0.05
    vocab_size = 50257 # 50k merges, 256 byte tokes, 1 special eof token
```

```python
gpt2_model = GPT2(GPTConfig())


model_state_dict = gpt2_model.state_dict()
total_params = sum(p.numel() for p in gpt2_model.parameters())  
print(total_params)
for k, v in model_state_dict.items():
    print(f"{k} =>  {v.shape}")

gpt2_model.to("mps")
```
