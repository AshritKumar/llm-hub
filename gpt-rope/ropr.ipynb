```python
import torch
hd = 4
freq_seq = torch.arange(hd)
print(freq_seq)
inv_freq = 1.0 / (10000 ** (freq_seq / hd))
print(inv_freq)
pos = torch.arange(5)
print(pos)

theta = torch.outer(pos, inv_freq)

print(theta)
print(theta.size())
x = theta.unsqueeze(0).unsqueeze(2)
print(x.size())

xx = torch.repeat_interleave(theta, 2, dim=-1)
yy = torch.repeat_interleave(x, 2, dim=-1)
print(xx.size())
print(yy)
```

```python
print(yy.size())
yy1 = yy[..., ::2]
yy2 = yy[..., 1::2]
print(yy1.size())
print(yy2.size())

yy3 = torch.stack((yy1, yy2), dim=-1)
print(yy3.size())
```
